## 4. Errors and interpretation

*Where is the model wrong? What does it lean on? A short error analysis beats a big metric table.*

Although the Random Forest performed best, it still makes mistakes. Some pages predicted to decline may actually represent seasonal or niche content, while some declining pages may not exhibit strong historical warning signals. The model appears to rely primarily on visibility and engagement features.

The most important features were:

days_with_impressions
log_impressions_90d
avg_position
content_age_days
char_count

These features indicate that search visibility, ranking performance, and content age contribute most to predicting declining pages. The model should therefore be used to prioritize review candidates rather than automatically refreshing content.
