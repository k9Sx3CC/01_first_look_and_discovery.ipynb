## 1. Ranked actions + reason codes

*The queue: what to do first, and why, in words a human trusts.*

he refresh queue ranks pages according to the model's predicted likelihood of decline. Pages with the highest refresh scores should be reviewed first because they are more likely to benefit from a content update. The queue is intended to help editors prioritize their workload rather than replace editorial judgment.

Each recommendation includes one or more reason codes that explain why a page was placed near the top of the queue. These reasons provide transparency and help reviewers understand the factors influencing the recommendation. Examples include:

*  declining_with_demand – the page is showing signs of decline while still receiving meaningful search demand.
*  low_ctr_visible_page – the page ranks in visible search positions but has a relatively *  low click-through rate, suggesting an opportunity to improve titles or metadata.
*  model_decline_risk – the model predicts an increased likelihood that the page is declining based on its historical search and engagement features.
*  visible_page – the page already receives search visibility, so improvements may have a measurable impact.
*  refresh_and_review_ctr – the page should be considered for both a content refresh and a review of its click-through performance.

These recommendations should be treated as decision-support signals. A human reviewer should confirm the recommendation before any content changes are made.

## 3. Human review + the no-go list

*What a person must check before acting. What should never be automated.*

Before refreshing any page, a reviewer should verify:

*  The page is actually losing traffic.
*  The topic is still relevant.
*  The decline is not caused by seasonality.
*  Content quality can realistically be improved.
*  Business priorities justify updating the page.

The following actions should not be automated:

*  Publishing rewritten content.
*  Deleting pages.
*  Redirecting URLs.
*  Changing business-critical information.
*  Making SEO decisions without human review.

*   Precision@50 decreases over time.
*   Search ranking behavior changes significantly.
