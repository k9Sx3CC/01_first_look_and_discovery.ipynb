## 4. Data limits

*What can this data never tell you? Unbalanced history, GSC-only early rows, window overlaps.*

This dataset contains observed page-level search and engagement metrics but cannot explain why a page's performance changes. External factors such as search engine algorithm updates, seasonality, competitor actions, and marketing campaigns are not included. Some fields also contain missing values, so predictions should be treated as decision support rather than evidence of causation.

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
import pandas as pd

df = pd.read_csv("data/raw/content_refresh_anonymized.csv")

print("Missing values per column:")
print(df.isnull().sum())

print("\nPercentage of missing values:")
print((df.isnull().mean() * 100).round(2))

print("\nLabel distribution:")
print(df["trend_direction"].value_counts(normalize=True).round(3))
```
