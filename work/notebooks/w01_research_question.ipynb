## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*
The starter dataset contains 30,000 anonymized content pages. Approximately 54.2% of pages are labeled as declining, providing enough positive examples for supervised learning. The correlation between search volume and actual impressions is approximately 0.001, indicating that keyword search volume alone is not a useful predictor of page traffic. Median word counts for declining and growing pages are very similar (about 2,909 vs. 2,848 words), suggesting that content length alone is unlikely to explain performance changes. These observations motivate using multiple behavioral and ranking features rather than relying on simple SEO assumptions.

## 4. Careful words: what I can and can't claim

*Write what your work will be able to say (observed, directional, decision-support) — and what it never will (causal proof, 'predicting Google').*
This project measures statistical relationships between observable SEO signals and future content decline.

The resulting model is intended as decision support for prioritizing content refreshes. The project can report observed patterns, measured prediction accuracy, and comparisons between baseline rules and machine learning models under honest validation. The project cannot prove why Google ranks pages the way it does, establish causal relationships, or claim to predict Google's ranking algorithm. Any conclusions are limited to the anonymized dataset and the evaluation methodology used.

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
import pandas as pd

df = pd.read_csv("data/raw/content_refresh_anonymized.csv")

print("="*50)
print("Dataset Summary")
print("="*50)
print(f"Rows: {len(df):,}")
print(f"Columns: {len(df.columns)}")

decline_rate = (df["trend_direction"]=="down").mean()
print(f"Declining pages: {decline_rate:.1%}")

corr = df["search_volume"].corr(df["impressions_90d"])
print(f"Search volume vs impressions correlation: {corr:.3f}")

ctr = (
    df[df["impressions_90d"]>=100]
    .groupby("position_tier")["ctr"]
    .mean()
    .sort_values(ascending=False)
)

print("\nAverage CTR by Position")
print(ctr.round(3))
```
